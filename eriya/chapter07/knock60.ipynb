```python
import pandas as pd

# ファイル名のリスト
files = ['train.tsv', 'dev.tsv']

for filename in files:
    # TSVファイルの読み込み
    df = pd.read_csv(filename, sep='\t')
    
    # ラベルごとの件数をカウント
    counts = df['label'].value_counts().sort_index()
    
    print(f'{filename}:')
    print(f'  Negative (0): {counts.get(0, 0)}')
    print(f'  Positive (1): {counts.get(1, 0)}\n')
```
