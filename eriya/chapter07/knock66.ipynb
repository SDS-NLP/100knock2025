```python
import pandas as pd
from collections import Counter
import joblib
from sklearn.metrics import confusion_matrix, classification_report

# dev.tsv を BoW 辞書付きで読み込む関数
def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')
    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])  # ラベルを文字列
        tokens = text.split()
        feature = dict(Counter(tokens))
        data.append({'text': text, 'label': label, 'feature': feature})
    return data

# モデルの読み込み
model = joblib.load('logreg_model.joblib')

# devデータの読み込みと特徴抽出
dev_data = load_and_vectorize('dev.tsv')
X_dev = [d['feature'] for d in dev_data]
y_true = [int(d['label']) for d in dev_data]

# 予測
y_pred = model.predict(X_dev)

# 混同行列を計算
cm = confusion_matrix(y_true, y_pred)
print("混同行列:")
print(cm)

# 分類レポート（精度・再現率・F1など）も表示したい場合
print("\n分類レポート:")
print(classification_report(y_true, y_pred, digits=4))
```
