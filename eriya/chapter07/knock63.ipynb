```python
import pandas as pd
from collections import Counter
import joblib

# モデル読み込み
model = joblib.load('logreg_model.joblib')

# dev.tsv 読み込みとBoW変換
def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')
    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])
        tokens = text.split()
        feature = dict(Counter(tokens))
        data.append({'text': text, 'label': label, 'feature': feature})
    return data

dev_data = load_and_vectorize('dev.tsv')

# 最初の事例で予測
first = dev_data[0]
X_first = [first['feature']]
y_true = first['label']
y_pred = model.predict(X_first)[0]

print("予測ラベル:", y_pred)
print("正解ラベル:", y_true)
print("一致しているか:", str(y_pred) == y_true)
```
