```python
import pandas as pd
from collections import Counter
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction import DictVectorizer
from sklearn.pipeline import make_pipeline

def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')
    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])  # ラベルは文字列
        tokens = text.split()
        feature = dict(Counter(tokens))
        data.append({'text': text, 'label': label, 'feature': feature})
    return data

# データ読み込み
train_data = load_and_vectorize('train.tsv')

# 特徴とラベルを抽出
X_train = [d['feature'] for d in train_data]
y_train = [int(d['label']) for d in train_data]  # ラベルは整数に変換

# BoW（Dict→ベクトル）＋ロジスティック回帰のパイプライン
model = make_pipeline(DictVectorizer(), LogisticRegression(max_iter=1000))

# モデルの学習
model.fit(X_train, y_train)

# 学習済みモデル（`model`）が得られる
# 例：最初の学習データの予測
print("予測:", model.predict([X_train[0]]))
print("正解:", y_train[0])
```

```python
import joblib

# 学習済みモデルを保存（例: 'logreg_model.joblib'）
joblib.dump(model, 'logreg_model.joblib')
```
