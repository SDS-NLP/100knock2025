```python
import joblib
import numpy as np
import pandas as pd

# 学習済みモデルの読み込み
model = joblib.load('logreg_model.joblib')

# DictVectorizer（BoWの語彙）と LogisticRegression モデルを取得
vectorizer = model.named_steps['dictvectorizer']
clf = model.named_steps['logisticregression']

# 単語（特徴名）と係数を取得
feature_names = vectorizer.get_feature_names_out()
weights = clf.coef_[0]  # shape: (1, n_features)

# DataFrameにまとめてソート
df_weights = pd.DataFrame({'feature': feature_names, 'weight': weights})
df_sorted = df_weights.sort_values('weight', ascending=False)

# 上位20語（ポジティブ寄り）・下位20語（ネガティブ寄り）を表示
print("★ 重みが高い（ポジティブ寄り）特徴トップ20:")
print(df_sorted.head(20).to_string(index=False))

print("\n★ 重みが低い（ネガティブ寄り）特徴トップ20:")
print(df_sorted.tail(20).to_string(index=False))
```
