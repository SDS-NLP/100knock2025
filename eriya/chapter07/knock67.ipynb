```python
import pandas as pd
from collections import Counter
import joblib
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


# dev.tsv を BoW 辞書付きで読み込む関数
def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')
    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])  # ラベルを文字列
        tokens = text.split()
        feature = dict(Counter(tokens))
        data.append({'text': text, 'label': label, 'feature': feature})
    return data

def evaluate(model, data, name="Data"):
    X = [d['feature'] for d in data]
    y_true = [int(d['label']) for d in data]
    y_pred = model.predict(X)

    # 評価指標の計算
    # 全体のうち、正しく分類された割合（正解の割合）
    acc = accuracy_score(y_true, y_pred)
    # モデルが「ポジティブ（1）」と予測したもののうち、実際に正しかった割合
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    print(f"\n{name}上での評価指標:")
    print(f"  正解率 (Accuracy):   {acc:.4f}")
    print(f"  適合率 (Precision):  {prec:.4f}")
    print(f"  再現率 (Recall):     {rec:.4f}")
    print(f"  F1スコア (F1):       {f1:.4f}")

# モデル読み込み
model = joblib.load('logreg_model.joblib')

# データ読み込み
train_data = load_and_vectorize('train.tsv')
dev_data = load_and_vectorize('dev.tsv')

# 評価
evaluate(model, train_data, name="学習データ")
evaluate(model, dev_data, name="検証データ")
```
