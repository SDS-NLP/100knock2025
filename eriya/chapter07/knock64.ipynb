```python
import pandas as pd
from collections import Counter
import joblib

# モデル読み込み
model = joblib.load('logreg_model.joblib')

# dev.tsv 読み込みとBoW変換（同じ関数を使ってOK）
def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')
    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])
        tokens = text.split()
        feature = dict(Counter(tokens))
        data.append({'text': text, 'label': label, 'feature': feature})
    return data

# 検証データの読み込み
dev_data = load_and_vectorize('dev.tsv')
first = dev_data[0]

# 特徴ベクトルの取り出し
X_first = [first['feature']]

# 条件付き確率を取得
probs = model.predict_proba(X_first)[0]

# 出力
print("テキスト:", first['text'])
print("正解ラベル:", first['label'])
print("予測確率:")
print(f"  Negative (0): {probs[0]:.4f}")
print(f"  Positive (1): {probs[1]:.4f}")
```
