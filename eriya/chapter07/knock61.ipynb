```python
import pandas as pd
from collections import Counter

def load_and_vectorize(filepath):
    df = pd.read_csv(filepath, sep='\t')

    data = []
    for _, row in df.iterrows():
        text = row['sentence']
        label = str(row['label'])  # ラベルを文字列に変換
        tokens = text.split()
        feature = dict(Counter(tokens))  # 出現頻度を辞書に
        data.append({'text': text, 'label': label, 'feature': feature})

    return data

# 学習データと検証データの読み込みと変換
train_data = load_and_vectorize('train.tsv')
dev_data = load_and_vectorize('dev.tsv')

# 例: 最初の事例を表示
print(train_data[0])
print(dev_data[0])
```
