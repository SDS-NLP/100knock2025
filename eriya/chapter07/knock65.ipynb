```python
import joblib
from collections import Counter

# ▼ ロジスティック回帰モデルを読み込む
model = joblib.load('logreg_model.joblib')

# ▼ 予測したいテキスト（任意に変更可能）
text = "the worst movie I've ever seen"

# ▼ 単語の頻度（BoW）を辞書形式で作成
tokens = text.split()
feature = dict(Counter(tokens))

# ▼ 予測（ラベルと確率）
pred_label = model.predict([feature])[0]
pred_proba = model.predict_proba([feature])[0]

# ▼ 出力
print("テキスト:", text)
print("予測ラベル:", pred_label)
print(f"  Negative (0): {pred_proba[0]:.4f}")
print(f"  Positive (1): {pred_proba[1]:.4f}")
```
