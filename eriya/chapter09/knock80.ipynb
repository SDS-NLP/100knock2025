```python
from transformers import BertTokenizer

# BERT-Base（uncased）トークナイザをロード
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

def tokenize_with_bert(sentence: str):
    """
    入力文をBERTのWordPieceトークナイザでトークン化し、
    [CLS], [SEP]トークンを含むトークン列を返します。
    """
    # encode_plus で special tokens を自動付与
    encoding = tokenizer.encode_plus(
        sentence,
        add_special_tokens=True
    )
    # トークンIDをトークン文字列に変換
    tokens = tokenizer.convert_ids_to_tokens(encoding['input_ids'])
    return tokens

# 使用例
sentence = "The movie was full of incomprehensibilities."
tokens = tokenize_with_bert(sentence)
print(tokens)
```
