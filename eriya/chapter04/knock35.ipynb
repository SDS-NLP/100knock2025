```python
import spacy
from spacy import displacy

# 日本語モデルをロード
try:
    nlp = spacy.load("ja_core_news_sm")
except OSError:
    print("日本語モデル 'ja_core_news_sm' が見つかりません。ダウンロードします。")
    spacy.cli.download("ja_core_news_sm")
    nlp = spacy.load("ja_core_news_sm")

text = "メロスは激怒した。"
doc = nlp(text)

# 係り受け木を可視化
# style='dep' で係り受け解析の可視化を指定
# options={'distance': 100} でトークン間の距離を調整し、見やすくする
# serve=True とするとブラウザで表示され、False の場合はHTMLが返される
displacy.render(doc, style="dep", options={'distance': 100}, jupyter=True)
```
