```python
import MeCab
import ipadic

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

CHASEN_ARGS = r' -F "%m\t%f[7]\t%f[6]\t%F-[0,1,2,3]\t%f[4]\t%f[5]\n"'
CHASEN_ARGS += r' -U "%m\t%m\t%m\t%F-[0,1,2,3]\t\t\n"'
tagger = MeCab.Tagger(ipadic.MECAB_ARGS + CHASEN_ARGS)
#tagger = MeCab.Tagger("-Odump" + ipadic.MECAB_ARGS)


# 結果を格納
verb_pairs = []

# テキスト解析
node = tagger.parseToNode(text)
while node:
    cols = node.feature.split(",")
    #print(cols)
    if cols[0] == "動詞":
        # 表層形、原形、品詞
        surface = node.surface
        base = cols[6]       # 原形
        verb_pairs.append((surface, base))
#    if len(cols) >= 3:
#        surface = cols[0]    # 表層形
#        base = cols[1]       # 原形
#        pos = cols[2]        # 品詞
#        if pos == "動詞":
#            verb_pairs.append((surface, base))
    node = node.next

# 表示
print("【動詞とその原型】")
for surface, base in verb_pairs:
    print(f"{surface} → {base}")
```
