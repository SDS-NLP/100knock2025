```python
import spacy

# 日本語モデルをロード
try:
    nlp = spacy.load("ja_core_news_sm")
except OSError:
    print("日本語モデル 'ja_core_news_sm' が見つかりません。ダウンロードします。")
    spacy.cli.download("ja_core_news_sm")
    nlp = spacy.load("ja_core_news_sm")

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

doc = nlp(text)

print("係り元\t係り先")
print("--------------------")

# 文節ごとに係り受けの関係を抽出
# spaCyでは、Token.headが係り先、Tokenが係り元となる
for token in doc:
    # 係り元と係り先が異なる場合のみ出力（自分自身への係り受けは除外）
    if token.head != token:
        print(f"{token.text}\t{token.head.text}")
```
