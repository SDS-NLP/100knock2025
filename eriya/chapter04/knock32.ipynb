```python
import MeCab
import ipadic

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

# MeCab設定
#CHASEN_ARGS = r' -F "%m\t%f[7]\t%f[6]\t%F-[0,1,2,3]\t%f[4]\t%f[5]\n"'
#CHASEN_ARGS += r' -U "%m\t%m\t%m\t%F-[0,1,2,3]\t\t\n"'
#tagger = MeCab.Tagger(ipadic.MECAB_ARGS + CHASEN_ARGS)

tagger = MeCab.Tagger(ipadic.MECAB_ARGS)
# ノードをすべてリスト化
nodes = []
print(nodes)
node = tagger.parseToNode(text)
while node:
    cols = node.feature.split("\t")
    if len(cols) >= 3:
        nodes.append({
            "surface": node.surface,
            "base": cols[1],
            "pos": cols[2]
        })
    node = node.next

# 「名詞 の 名詞」パターンを抽出
noun_possession = []
for i in range(len(nodes) - 2):
    a, b, c = nodes[i], nodes[i + 1], nodes[i + 2]
    if a["pos"] == "名詞" and b["surface"] == "の" and c["pos"] == "名詞":
        noun_possession.append(f'{a["surface"]}の{c["surface"]}')

print("名詞の所有格:", noun_possession)
```
