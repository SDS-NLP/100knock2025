```python
import fugashi
from collections import Counter
import os

def analyze_kokoro_nouns(file_path="kokoro.txt", top_n=20):
    """
    指定されたテキストファイルを読み込み、形態素解析を行い、
    名詞の出現頻度を計算して上位N語を表示します。
    """
    if not os.path.exists(file_path):
        print(f"エラー: ファイル '{file_path}' が見つかりません。")
        print("ファイルパスが正しいか、ファイルが同じディレクトリにあるか確認してください。")
        return

    # Fugashi (MeCab) タガーを初期化
    # デフォルトでは unidic-lite が使われることが多いです。
    tagger = fugashi.Tagger()

    all_nouns = []
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            text = f.read()

        # 形態素解析を実行し、各トークンの品詞情報を取得
        for word in tagger(text):
            # 品詞が「名詞」であるかを確認
            # Unidicの場合、品詞はword.posで取得でき、'名詞'で始まるものが名詞
            if word.pos.startswith('名詞'):
                all_nouns.append(word.surface) # word.surface は単語の表層形

        # 名詞の出現頻度を計算
        noun_counts = Counter(all_nouns)

        print(f"--- '{file_path}' 内の名詞の出現頻度の高い上位 {top_n} 語 ---")
        # 出現頻度の高い順にN語を表示
        for noun, count in noun_counts.most_common(top_n):
            print(f"'{noun}': {count} 回")

    except Exception as e:
        print(f"ファイル処理中にエラーが発生しました: {e}")

# 分析の実行
if __name__ == "__main__":
    analyze_kokoro_nouns("kokoro.txt")
```
