```python
#!pip install mecab-python3
import MeCab
from collections import Counter
#!pip install pandas
import pandas as pd

# テキスト読み込み
with open('kokoro.txt', 'r', encoding='utf-8') as f:
    text = f.read()

# MeCabで形態素解析
tagger = MeCab.Tagger('-Ochasen')
node = tagger.parseToNode(text)

# 名詞・動詞・形容詞・副詞の基本形抽出
words = []
while node:
    features = node.feature.split(',')
    pos = features[0]
    base = features[6]
    if base != '*' and pos in ['名詞', '動詞', '形容詞', '副詞']:
        words.append(base)
    node = node.next

# 出現頻度計算
counter = Counter(words)

# 上位20語を表示
df_top20 = pd.DataFrame(counter.most_common(20), columns=['単語', '出現頻度'])
tools.display_dataframe_to_user('kokoro_top20', df_top20)
```

```python
!pip install fugashi unidic-lite
import fugashi
from collections import Counter
import os

def analyze_kokoro_text(file_path="kokoro.txt", top_n=20):
    """
    指定されたテキストファイルを読み込み、形態素解析を行い、
    単語の出現頻度を計算して上位N語を表示します。
    """
    if not os.path.exists(file_path):
        print(f"エラー: ファイル '{file_path}' が見つかりません。")
        print("ファイルパスが正しいか、ファイルが同じディレクトリにあるか確認してください。")
        return

    # Fugashi (MeCab) タガーを初期化
    # ipadicを使用する場合は fugashi.Tagger('-Owakati')
    # unidicを使用する場合は fugashi.Tagger('-Owakati -d /path/to/unidic') のように指定
    # デフォルトでは unidic-lite が使われることが多いです。
    tagger = fugashi.Tagger()

    all_words = []
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            text = f.read()

        # 形態素解析を実行
        # tagger.parse()は文字列を返し、単語がスペース区切りで並ぶ
        # tagger.parseToNode()や tagger.parse()の結果を処理して、品詞フィルタリングも可能
        # 今回はシンプルに分かち書きされた単語を抽出
        words = tagger.parse(text).strip().split()
        
        # 不要な記号や助詞などを除外する場合のフィルタリング例 (必要に応じて追加・調整)
        # ただし、今回は全ての単語を対象とします。
        # filtered_words = [word for word in words if word.isalnum()] # 英数字のみ
        
        all_words.extend(words)

        # 単語の出現頻度を計算
        word_counts = Counter(all_words)

        print(f"--- '{file_path}' 内の出現頻度の高い上位 {top_n} 語 ---")
        # 出現頻度の高い順にN語を表示
        for word, count in word_counts.most_common(top_n):
            print(f"'{word}': {count} 回")

    except Exception as e:
        print(f"ファイル処理中にエラーが発生しました: {e}")

# 分析の実行
if __name__ == "__main__":
    analyze_kokoro_text("kokoro.txt")
```
