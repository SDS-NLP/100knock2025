```python
import MeCab

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

# MeCabのTaggerを用意
#tagger = MeCab.Tagger()

#import MeCab

# mecabrc のパスをフルで指定（例: 実際のパスに置き換えてください）
#mecabrc_path = r"C:/Users/eriya/anaconda3/envs/python-env/Lib/site-packages/mecab-0.996.11.dist-info"
#tagger = MeCab.Tagger(f"-r {mecabrc_path}")

#import MeCab
import ipadic

#CHASEN_ARGS = r' -F "%m\t%f[7]\t%f[6]\t%F-[0,1,2,3]\t%f[4]\t%f[5]\n"'
#CHASEN_ARGS += r' -U "%m\t%m\t%m\t%F-[0,1,2,3]\t\t\n"'
#tagger = MeCab.Tagger(ipadic.MECAB_ARGS + CHASEN_ARGS)
tagger = MeCab.Tagger(ipadic.MECAB_ARGS)
#print(tagger.parse("図書館にいた事がバレた"))

# 出力を格納するリスト
verbs = []

# 形態素解析
node = tagger.parseToNode(text)
while node:
    features = node.feature.split(',')
    if features[0] == "動詞":
        verbs.append(node.surface)  # 活用形をそのまま取得
    node = node.next

print("動詞（表層形）:", verbs)
```
