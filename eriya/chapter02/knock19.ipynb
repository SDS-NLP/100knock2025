```python
def sort_by_third_desc():
    # 入力ファイル、出力ファイルを直接指定
    input_path  = "popular-names.txt"
    output_path = "sorted.txt"

    # ファイル読み込み＋ (行, count) タプルのリスト化
    records = []
    with open(input_path, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.rstrip('\n').split('\t')
            if len(parts) < 3:
                continue
            try:
                count = int(parts[2])
            except ValueError:
                continue
            records.append((line, count))

    # 3列目の数値で降順ソート
    records.sort(key=lambda x: x[1], reverse=True)

    # ソート結果を書き出し
    with open(output_path, 'w', encoding='utf-8') as f_out:
        for line, _ in records:
            f_out.write(line)

    print(f"Sorted output written to {output_path}")

if __name__ == "__main__":
    sort_by_third_desc()



### sort -t $'\t' -k3,3nr popular-names.txt > sorted.txt
```
