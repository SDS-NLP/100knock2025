```python
from gensim.models import KeyedVectors

# モデルの読み込み（パスを環境に応じて変更）
model = KeyedVectors.load_word2vec_format("GoogleNews-vectors-negative300.bin.gz", binary=True)

# 入力ファイル（capital-common-countries を含む questions-words.txt）
input_file = "questions-words.txt"

in_target_section = False  # capital-common-countries セクションのフラグ

with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        if line.startswith(":"):
            # セクション判定
            in_target_section = (line.strip() == ": capital-common-countries")
            continue
        if not in_target_section:
            continue  # 他のセクションは無視

        try:
            a, b, c, *_ = line.split()
            # アナロジーベクトル: b - a + c
            result = model.most_similar(positive=[b, c], negative=[a], topn=1)
            d, similarity = result[0]
            print(f"{a:15} {b:15} {c:15} {d:15} {similarity:.4f}")
        except KeyError as e:
            print(f"{a:15} {b:15} {c:15} {'N/A':15} {'0.0000'}  # {str(e)}")
```
