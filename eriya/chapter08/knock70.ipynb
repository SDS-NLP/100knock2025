```python
import numpy as np
from gensim.models import KeyedVectors
# GoogleNews の pretrained Word2Vec のロード（300次元）
# ※ファイル: GoogleNews-vectors-negative300.bin.gz（事前ダウンロード必要）
word2vec_path = "C:/Users/eriya/Documents/classcontents/25春夏/100knock2025/chapter06/GoogleNews-vectors-negative300.bin.gz"
model = KeyedVectors.load_word2vec_format(word2vec_path, binary=True)

embedding_dim = model.vector_size        # 例：300
vocab = list(model.key_to_index.keys())  # 語彙リスト（デフォルト：全300万語）
max_vocab_size = 50000                   # 必要に応じて制限

# <PAD>用に1行追加（全て0）
embedding_matrix = [np.zeros(embedding_dim)]  # 0番目にゼロベクトル
token_to_id = {'<PAD>': 0}
id_to_token = {0: '<PAD>'}

# 語彙の構築（2行目以降に単語ベクトルを格納）
for i, word in enumerate(vocab[:max_vocab_size], start=1):
    vector = model[word]
    embedding_matrix.append(vector)
    token_to_id[word] = i
    id_to_token[i] = word

embedding_matrix = np.array(embedding_matrix)  # numpy配列に変換

print("埋め込み行列の形状:", embedding_matrix.shape)
print("例: 'good' → ID:", token_to_id.get('good'))
print("例: ID 1 → 単語:", id_to_token.get(1))
```

```python
np.save('embedding_matrix.npy', embedding_matrix)
```

```python
import numpy as np
import gensim

# 事前学習済みGoogle Newsベクトルの読み込み
w2v_path = 'C:/Users/eriya/Documents/classcontents/25春夏/100knock2025/chapter06/GoogleNews-vectors-negative300.bin.gz'
w2v = gensim.models.KeyedVectors.load_word2vec_format(w2v_path, binary=True)

# 語彙を制限（例: 上位100,000語）
vocab_list = list(w2v.key_to_index.keys())[:100000]
vocab_size = len(vocab_list) + 1  # +1 for <PAD>
emb_dim = w2v.vector_size

# 埋め込み行列 & 双方向辞書の初期化
embedding_matrix = np.zeros((vocab_size, emb_dim), dtype=np.float32)
token2id = {'<PAD>': 0}
id2token = {0: '<PAD>'}

for idx, token in enumerate(vocab_list, start=1):
    token2id[token] = idx
    id2token[idx]   = token
    embedding_matrix[idx] = w2v[token]

# 保存
to_save = {'embedding_matrix': embedding_matrix, 'token2id': token2id, 'id2token': id2token}
np.save('70_embeddings.npy', embedding_matrix)
import pickle
with open('70_token_maps.pkl', 'wb') as f:
    pickle.dump({'token2id': token2id, 'id2token': id2token}, f)
```
