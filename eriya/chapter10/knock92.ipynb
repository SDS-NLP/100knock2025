```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch
import numpy as np

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
model.eval()

prompt = "The movie was full of"
input_ids = tokenizer.encode(prompt, return_tensors="pt")

output = model.generate(input_ids, max_length=15, do_sample=False)
decoded = tokenizer.decode(output[0], skip_special_tokens=True)
print("Generated:", decoded)

log_probs = []
with torch.no_grad():
    for i in range(1, output.shape[1]):
        input_slice = output[:, :i]
        logits = model(input_slice).logits
        probs = torch.softmax(logits[:, -1, :], dim=-1)
        token_id = output[0, i]
        prob = probs[0, token_id].item()
        log_probs.append((tokenizer.decode([token_id]), np.log(prob)))

for token, log_p in log_probs:
    print(f"Token: {token!r}, log-prob: {log_p:.4f}")
```
